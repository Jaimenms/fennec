# Example 2 - Solar house with EV

```python
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt

from fennec.fennec import run
from fennec.utils import get_detailed_results
from fennec.examples.example_02 import get_data
```

```python
data = get_data()
```

```python
results, instance = run(data, time_limit=120, mip_abs_gap = 0.01, mip_rel_gap=0.01)
instance.solutions.load_from(results)
```

```python
print("Status:", results.solver.status)
```

```python
detailed_results = get_detailed_results(results, instance)
df_parameters = pd.DataFrame(detailed_results["Parameters"])
df_variables = pd.DataFrame(detailed_results["Variables"])
df_sets = pd.DataFrame(detailed_results["Sets"])
```

```python
def get_set(df, name):
    return df[df['Set'] == name]['Value'].values
def get_variable(df, name):
    return df[df['Variable'] == name]['Value'].values
def get_parameter(df, name):
    return df[df['Parameter'] == name]['Index'].values, df[df['Parameter'] == name]['Value'].values
```

```python
t = get_set(df_sets, 't')
grid_to = get_variable(df_variables, 'grid.energy_to')
grid_from = get_variable(df_variables, 'grid.energy_from')
_, tariff_to = get_parameter(df_parameters, 'grid.tariff_energy_from')
_, tariff_from = get_parameter(df_parameters, 'grid.tariff_energy_to')
_, load_to = get_parameter(df_parameters, 'node[LOAD01].energy_to')
_, load_from = get_parameter(df_parameters, 'node[LOAD01].energy_from')
_, pv_to = get_parameter(df_parameters, 'node[PV01].energy_to')
_, pv_from = get_parameter(df_parameters, 'node[PV01].energy_from')
_, is_plugged = get_parameter(df_parameters, 'vehicle[EV01].is_plugged')

soc = get_variable(df_variables, 'vehicle[EV01].soc')
charge = get_variable(df_variables, 'vehicle[EV01].charge')
energy_to = get_variable(df_variables, 'vehicle[EV01].energy_to')
external_to = get_variable(df_variables, 'vehicle[EV01].external_to')
external_from = get_variable(df_variables, 'vehicle[EV01].external_from')
i_soc_back_home, soc_back_home = get_parameter(df_parameters, 'vehicle[EV01].soc_back_home')
i_soc_objective, soc_objective = get_parameter(df_parameters, 'vehicleController[CONT01].soc_objective')
```

```python
mpl.rcParams.update({
    'figure.figsize': (12, 10),   # default figure size
    'axes.grid': True,          # turn on grid by default
    'axes.titleweight': 'bold', # example of making titles bold
    'lines.linewidth': 2.0,     # default line width
    'lines.linestyle': '-',     # default line width
    'lines.markersize': 3.0,
    'font.size': 8,            # base font size
    'axes.labelsize': 8,       # axis label font size
    'axes.titlesize': 8,       # axis title font size
})
```

```python
fig1, axs = plt.subplots(8, 1, sharex=True)

axs[0].step(t, grid_to, where='post', marker='o', linestyle='-', color = 'red')
axs[0].step(t, grid_from, where='post', marker='o', linestyle='-', color = 'green')
axs[0].set_ylabel("Energy (kWh)")
axs[0].set_title("Grid")
axs[0].legend(["to", "from"])

axs[1].step(t, tariff_to, where='post', marker='o', linestyle='-', color = 'red')
axs[1].step(t, tariff_from, where='post', marker='o', linestyle='-', color = 'green')
axs[1].set_ylabel("Tariff ($/kWh)")
axs[1].set_title("Grid")
axs[1].legend(["to", "from"])

axs[2].step(t, load_to, where='post', marker='o', linestyle='-', color = 'red')
axs[2].step(t, load_from, where='post', marker='o', linestyle='-', color = 'green')
axs[2].set_ylabel("Energy (kWh)")
axs[2].set_title("Load")
axs[2].legend(["to", "from"])

axs[3].step(t, pv_to, where='post', marker='o', linestyle='-', color = 'red')
axs[3].step(t, pv_from, where='post', marker='o', linestyle='-', color = 'green')
axs[3].set_ylabel("Energy (kWh)")
axs[3].set_title("PV")
axs[3].legend(["to", "from"])

axs[4].step(t, soc, where='post', marker='o', linestyle='-', color = 'blue')
axs[4].step(i_soc_objective, soc_objective, where='post', marker='o', linestyle='-', color = 'pink', markersize=10, linewidth=0)
axs[4].step(i_soc_back_home, soc_back_home, where='post', marker='o', linestyle='-', color = 'purple', markersize=10, linewidth=0)
axs[4].set_ylabel("SOC (%)")
axs[4].set_title("SOC")
axs[4].legend(["value", "objectives", "back home"])

axs[5].step(t, charge,  where='post', marker='o', linestyle='-', color = 'red')
axs[5].set_ylabel("-")
axs[5].set_title("Battery")
axs[5].legend(["charged"])

axs[6].step(t, is_plugged,  where='post', marker='o', linestyle='-', color = 'red')
axs[6].set_ylabel("-")
axs[6].set_title("Is plugged")
axs[6].legend(["charged"])

axs[7].step(t, external_to,  where='post', marker='o', linestyle='-', color = 'red')
axs[7].step(t, external_from,  where='post', marker='o', linestyle='-', color = 'green')
axs[7].step(t, energy_to,  where='post', marker='o', linestyle='-', color = 'blue')
axs[7].set_ylabel("Energy (kWh)")
axs[7].set_title("Battery")
axs[7].legend(["External To", "External From", "Charge"])

```
